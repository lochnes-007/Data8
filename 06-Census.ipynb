```python
full = Table.read_table("nc-est2019-agesex-res.csv")
```

```python
full.show(5)
```

```python
partial = full.select("AGE", "SEX", "POPESTIMATE2014", "POPESTIMATE2019")
partial.show(5)
```

```python
us_pop = partial.relabeled("POPESTIMATE2014", "2014").relabeled("POPESTIMATE2019", "2019")
us_pop.show(5)
```

```python
us_pop.where("AGE", are.above_or_equal_to(80)).sort("AGE").select(1,0,2,3)
```
