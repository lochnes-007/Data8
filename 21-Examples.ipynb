```python
from datascience import *
import numpy as np

import matplotlib.pyplot as plt

%matplotlib inline
```

```python
def simulate_one_count(sample_size):
    return np.count_nonzero(np.random.choice(['H', 'T'], sample_size)=='H')
simulate_one_count(100)
```

```python
num_sims = 1000
counts = make_array()

for i in range(num_sims):
    counts = np.append(counts, simulate_one_count(200))

# counts
```

```python
# P_val 5%

Table().with_column("Super Sode Dist", counts).hist()
```

```python
(np.sum(counts <= 91) / num_sims)*100
```
