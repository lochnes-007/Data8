```python
def my_round(mr, yr):
    if(mr > yr):
        return 1
    elif(mr==yr):
        return 0
    elif(mr<yr):
        return -1
```

```python
die_faces = np.arange(1, 7)
die_faces
```

```python
def simulate_one_round():
    mr = np.random.choice(die_faces)
    yr = np.random.choice(die_faces)
    return my_round(mr, yr)
```
