```python
top_movies = Table.read_table("top_movies_2017.csv")
top_movies.show(3)
```

```python
studio_counts = top_movies.group("Studio")
studio_counts
```

```python
studio_counts.sort("count").barh("Studio", "count")
```

```python
studio_counts.with_column("Percent", studio_counts.column("count") / sum(studio_counts.column("count")) * 100).sort("Percent", descending=True).barh("Studio", "Percent")
```
