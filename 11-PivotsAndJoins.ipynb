```python
sky = Table.read_table("skyscrapers.csv")
sky = sky.with_column("age", 2026 - sky.column("completed")).drop("completed")
sky.show(5)
```

```python
sky_g = sky.select("name", "city", "material", "height", "age").group(["city", "material"], collect=max)
sky_g
```

```python
sky_p = sky.pivot("material", "city", "height", np.max)
sky_p
```

```python
sky_p_withconc_steel_diff = sky_p.with_column("diffconst", abs(sky_p.column("steel") - sky_p.column("concrete")))
```

```python
sky_p_withconc_steel_diff
```

```python
def take_the_first_one(a):
    return a.item(0)

sky.sort("age", descending=True).pivot("material", "city", "name",take_the_first_one)
```
