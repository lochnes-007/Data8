```python
from datascience import *
import numpy as np


import matplotlib.pyplot as plt

%matplotlib inline
```

```python
# Simulating heads in 100 coin tosses
coin = ["h", "t"]
def num_heads(n):
    return sum(np.random.choice(coin, n)=="h")
```

```python
reps = 10000

num_heads(reps)
```

```python
outcomes = make_array()

for i in range(reps):
    outcomes = np.append(outcomes, num_heads(100))
```

```python
np.sum(outcomes)
```

```python
heads = Table().with_column("heads", outcomes)
heads.show(5)
```
