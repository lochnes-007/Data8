```python
from datascience import *
import numpy as np
```

```python
nba_pg_2020 = Table.read_table("nba_salaries.csv").where("season", 2020).drop("season")
```

```python
nba_pg_2020
```

```python
warriors = (
    nba_pg_2020
    .where("team", "Golden State Warriors")
    .drop("team")
    .relabeled("salary", "$"))
warriors.show(3)
```

```python
salary = warriors.column("$")
```

```python
salary.sum()
```

```python
salary.max()
```
